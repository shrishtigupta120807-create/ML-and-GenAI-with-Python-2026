```python
#que 1
def sum(n):
    sum=n*(n+1)/2
    print(f"sum is:{sum}")
sum(6)
```

```python
#que 2
def factorial(num):
    if(num==0):
        return 1
    return num*factorial(num-1)
print(factorial(4))
```

```python
#que 3
def fibonacci(num):
    if(num==1):
      return 0
    if(num==2):
      return 1
    return fibonacci(num-1)+fibonacci(num-2)
print(fibonacci(6))
```
