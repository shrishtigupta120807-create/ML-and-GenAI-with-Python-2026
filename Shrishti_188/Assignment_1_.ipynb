```python
#que 1
def area(length,breadth):
 area=length*breadth
 print(f"area is:{area}")
area(4,5)
```

```python
#que 2
def simple_interest(principal,time,rate):
 simple_interest=principal*time*rate/100
 print(f"simple_interest:{simple_interest}")
simple_interest(100,2,3)
```

```python
#que 3
def conversion(celcius):
 farhrenheit=celcius*9/5+32
 print(f"temp in farhenheit:{farhrenheit}")
conversion(45)
```

```python
#que 4
def avg(a,b,c):
 avg=a+b+c/3
 print(f"avg is:{avg}")
avg(4,5,9)
```

```python
#que 5
def square_cube(num):
 square=num*num
 cube=num*num*num
 print(f"square is:{square}")
 print(f"cube is:{cube}")
square_cube(6)
```

```python
#que 6
def swap(a,b):
  a,b=b,a
  print(f"a is:{a}")
  print(f"b is:{b}")
swap(4,5)
```

```python
#que 7
def student_report():
 phy_marks=int(input("Enter physics marks"))
 maths_marks=int(input("Enter maths marks"))
 ds_marks=int(input("Enter ds marks"))
 sspd_marks=int(input("Enter sspd marks"))
 ids_marks=int(input("Enter ids marks"))
 csa_marks=int(input("Enter csa marks"))
 total_marks=phy_marks+maths_marks+ds_marks+sspd_marks+ids_marks+csa_marks
 print(f"total_marks:{total_marks}")
 percentage=(total_marks/600)*100
 print(f"percentage is:{percentage}")

student_report()
```
